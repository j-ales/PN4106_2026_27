## 4. Variables

A **variable** gives a value a name. The assignment operator `=` stores the value on its right under the name on its left.

Read `score = 8` as “assign the value 8 to the name `score`.” It is an instruction, not a mathematical equation.

````python
score = 8
print(score)
````

We can use the variable in later calculations.

````python
score + 2
````

Running another assignment can replace the value stored under the same name.

````python
score = 10
print(score)
````

## 5. Simple types

Python stores different kinds of values. Three common types are:

- `str`: text, written inside quotation marks;
- `int`: a whole number;
- `float`: a number with a decimal point.

The `type()` function tells us the type of a value.

````python
module_name = "PN4106"
week_number = 1
example_decimal = 2.5
````

````python
print(module_name)
print(week_number)
print(example_decimal)
````

````python
print(type(module_name))
print(type(week_number))
print(type(example_decimal))
````

Quotation marks matter. `"10"` is text, whereas `10` is a number. Their types are therefore different.

````python
print(type("10"))
print(type(10))
````

In python, strings and integers cannot be combined directly.  For example you will get an error if you try to run this code:

````python
answer = 10 + "10"
````

## 6. Lists

A **list** stores a sequence of values together. Lists are fundamental in Python and will also be useful when we begin working with pandas later in the module.

We create a list using square brackets `[]`, with commas between the items.

````python
colors = ["red", "green", "blue"]
print(colors)
````

The items have positions called **indexes**. Python starts counting at zero, so index `0` selects the first item. Index `-1` is a convenient way to select the final item.

````python
print(colors[0])
print(colors[1])
print(colors[-1])
````

The `len()` function reports how many items are in a list.

````python
len(colors)
````

Lists can be changed. The `.append()` method adds one new item to the end.

````python
colours.append("yellow")
print(colors)
print(len(colors))
````
