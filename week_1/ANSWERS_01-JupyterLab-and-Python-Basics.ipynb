## 5. Variables

Here is one possible solution. Your name and number of available hours may be different.

```python
student_name = "Alex"
week_number = 1
hours_available = 2.5
```

The assignment operator `=` stores each value under the name on its left. Text needs quotation marks; numeric values do not.

```python
print(student_name)
print(week_number)
print(hours_available)
```

We pass each variable to `print()` by placing its name inside the parentheses. Do not put quotation marks around the variable name: `print("student_name")` would print the literal text rather than the value stored in the variable.

```python
print(type(student_name))
print(type(week_number))
print(type(hours_available))
```

The types are `str`, `int`, and `float`. A string is text, an integer is a whole number, and a float is a number containing a decimal point.

## 6. Lists

Here is one possible list. The square brackets create the list, commas separate its items, and quotation marks identify each item as text.

```python
snacks = ["apple", "popcorn", "chocolate"]
print(snacks)
```

Python starts counting list positions at zero, so `[0]` selects the first item.

```python
print(snacks[0])
```

The index `-1` selects the final item without requiring us to know how long the list is.

```python
print(snacks[-1])
```

The `len()` function returns the number of items. This list initially contains three items.

```python
print(len(snacks))
```

The `.append()` method adds one item to the end of the existing list. After appending `"crisps"`, the list contains four items.

```python
snacks.append("crisps")
print(snacks)
print(len(snacks))
```

A common mistake is to write `snacks[1]` when trying to obtain the first item. Index `1` is the second item because counting begins at zero.
