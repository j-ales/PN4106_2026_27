## Core exercise 7: Two kinds of debugging

Here's a reminder of what to do when something is wrong:

1. **Expect:** State what the code should do.
2. **Classify:** Does it fail to run, or does it run with the wrong result?
3. **Read:** If there is a traceback, read it from the bottom.
4. **Inspect:** Check relevant values using `print()`, `type()`, and `len()`.
5. **Change:** Change one thing for a reason.
6. **Check:** Run the code again and compare the result with your expectation.

### Debugging B

**Expected Behaviour:** The code is intended to average the 3 highest values.

Expected Output:
```Average of the three highest scores: 75.0```


````python
scores = [52, 68, 74, 61, 83]

ordered_scores = sorted(scores, reverse=True)
top_three = ordered_scores[:3]

total_top_three = sum(top_three)
average_top_three = total_top_three / len(scores)

print("Average of the three highest scores:", average_top_three)
````
