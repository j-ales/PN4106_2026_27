## Core exercise 5: Build a Boolean mask

Only 72 and 83 are at least 60, so only their positions contain `True`.

```python
scores = [45, 72, 51, 83]
keep = [False, True, False, True]

print(scores)
print(keep)
```

```python
print(len(scores))
print(len(keep))
print(len(scores) == len(keep))
```

Both lists have four items. This one-to-one correspondence is required when a Boolean mask selects rows from a pandas table.

## Core exercise 7: Two kinds of debugging

The same routine applies to both examples: establish the expected result, classify the problem, inspect the evidence, change one thing, and check again.

### Debugging A

**Expected:** The code should print `25`.

**Classification:** The code does not run.

**Evidence:** The final traceback line reports a `TypeError`. Inspecting the variable shows that `participant_count` is a string. Python cannot add the integer `5` to the string `"20"`.

**Change:** Store 20 as an integer by removing the quotation marks.

**Check:** The corrected code prints `25`.

```python
# Inspect the value before changing it
participant_count = "20"
print(participant_count)
print(type(participant_count))
```

```python
# Corrected code
participant_count = 20
new_count = participant_count + 5
print(new_count)
```

### Debugging B

**Expected:** The code should calculate the average of the three highest scores and print `Average of the three highest scores: 75.0`.

**Classification:** The code runs but gives the wrong result.

**Evidence:** It prints `Average of the three highest scores: 45.0` and produces no traceback.

The intermediate values help us locate the problem:

- `ordered_scores` is `[83, 74, 68, 61, 52]`, so the sorting is correct.
- `top_three` is `[83, 74, 68]`, so the slice is correct.
- `total_top_three` is `225`, so the total is correct.
- `len(scores)` is 5, but `len(top_three)` is 3.

**Cause:** The final calculation divides by the number of values in the original list rather than the number of selected values being averaged. The bug is downstream: the earlier steps are working correctly.

**Change:** Replace `len(scores)` with `len(top_three)`.

**Check:** The corrected code prints the expected average, `75.0`.

```python
# Inspect the intermediate results
scores = [52, 68, 74, 61, 83]
ordered_scores = sorted(scores, reverse=True)
top_three = ordered_scores[:3]
total_top_three = sum(top_three)

print(ordered_scores)
print(top_three)
print(total_top_three)
print(len(scores))
print(len(top_three))
```

```python
# Corrected code
scores = [52, 68, 74, 61, 83]

ordered_scores = sorted(scores, reverse=True)
top_three = ordered_scores[:3]

total_top_three = sum(top_three)
average_top_three = total_top_three / len(top_three)

print("Average of the three highest scores:", average_top_three)
```

## Stretch goal 2: Harder debugging challenge

This problem reveals three issues in sequence. Fixing the first problem does not mean the entire program is correct.

### Problem 1: Mixed types

**Classification:** Does not run.

**Evidence:** `sorted()` reports a `TypeError` because Python cannot order the string `"78"` alongside integers.

**Change:** Replace `"78"` with the integer `78`.

After this change, running the cell reveals another problem.

### Problem 2: Incorrect variable name

**Evidence:** The final traceback line reports a `NameError`: `top_scores` has not been defined. The result was stored under the name `top_three`.

**Change:** Print `top_three`.

The code now runs, but its output still needs to be checked.

### Problem 3: Incorrect slice

**Classification:** Runs but gives the wrong result.

**Evidence:** The output contains only `[92, 84]`, while three values were expected.

The completed code is:

```python
scores = [61, 78, 55, 92, 70, 84]

ordered_scores = sorted(scores, reverse=True)
top_three = ordered_scores[0:3]

print("Top three:", top_three)
```

The important debugging lesson is the sequence: repair one evidenced problem, rerun the code, and then reassess. A program can contain more than one problem, and code that finally runs can still produce the wrong result.
